## Subsección 4: Ramas y viaje por las versiones del repositorio (Git)

Hasta ahora usted ha hecho **un commit por cada punto**. Cada commit es una **fotografía** (*snapshot*) completa del repositorio en un momento dado, identificada por un código único llamado **hash** (por ejemplo `a3f9c1e`). En esta subsección aprenderá a:

1. Trabajar en una **rama** (*branch*) sin afectar la versión principal del proyecto.
2. **Unir** (*merge*) esa rama con la rama principal.
3. **Viajar** por las diferentes versiones del repositorio: ver, comparar, recuperar y deshacer cambios.

> **Importante:** los comandos de Git se escriben en la **terminal** de Codespaces (`Ctrl + ñ` o menú *Terminal → New Terminal*). Los comandos que **solo consultan** información (`git log`, `git show`, `git diff`, `git status`) también pueden ejecutarse desde una celda del cuaderno anteponiendo `!`, por ejemplo `!git log --oneline`. Los comandos que **cambian de versión** (`git switch`, `git checkout`, `git merge`, `git revert`) ejecútelos **siempre desde la terminal**, porque modifican el archivo del cuaderno que usted tiene abierto.

### 4.1 ¿Qué es una rama?

Una **rama** es simplemente un **puntero** con nombre que apunta a un commit. La rama principal suele llamarse `main`. Cuando usted crea una rama nueva, Git no copia archivos: solo crea un nuevo puntero que avanzará con los commits que haga en ella, mientras `main` se queda quieta.

`HEAD` es otro puntero especial que indica **en qué lugar está usted parado** en este momento (normalmente, en la punta de una rama).

```
                         C4 ── C5      ← analisis-extra  (HEAD)
                        /
C0 ── C1 ── C2 ── C3 ──┘               ← main
```

Después de unir (`merge`) la rama `analisis-extra` en `main`, ambas historias quedan juntas:

```
C0 ── C1 ── C2 ── C3 ── C4 ── C5       ← main, analisis-extra
```

**¿Para qué sirven las ramas?** Para experimentar, desarrollar una funcionalidad nueva o resolver un problema **sin dañar** la versión estable. Si el experimento sale mal, basta con abandonar la rama; si sale bien, se une a `main`.

| Comando | Qué hace |
|---|---|
| `git branch` | Lista las ramas locales (la actual aparece con `*`). |
| `git branch -a` | Lista también las ramas remotas (las de GitHub). |
| `git switch -c <rama>` | Crea una rama nueva y se cambia a ella (equivale a `git checkout -b <rama>`). |
| `git switch <rama>` | Se cambia a una rama existente. |
| `git push -u origin <rama>` | Sube la rama a GitHub por primera vez. |
| `git merge <rama>` | Une la rama indicada **dentro de la rama actual**. |
| `git branch -d <rama>` | Borra una rama local que ya fue unida. |

> En Codespaces también puede ver y cambiar de rama haciendo clic en el nombre de la rama en la **esquina inferior izquierda** de la ventana, o desde el panel **Source Control**.

### 4.3 Viajar por las versiones del repositorio

Como cada commit es una fotografía completa, Git permite **volver a cualquier punto de la historia**. Estas son las herramientas principales, de la más inofensiva a la más peligrosa:

| Comando | Qué hace | ¿Modifica la historia? |
|---|---|---|
| `git log --oneline` | Lista los commits (hash corto + mensaje). | No |
| `git show <hash>` | Muestra qué cambió en un commit concreto. | No |
| `git diff <hash1> <hash2>` | Compara dos versiones. | No |
| `git switch --detach <hash>` | **Viaja** a un commit antiguo para verlo (modo *detached HEAD*). | No |
| `git switch main` | Regresa al presente. | No |
| `git restore --source=<hash> <archivo>` | Trae **un archivo** tal como estaba en ese commit. | No (queda como cambio pendiente) |
| `git revert <hash>` | Crea un **commit nuevo** que deshace lo que hizo `<hash>`. | No, **agrega** historia (seguro) |
| `git reset --hard <hash>` | Mueve la rama hacia atrás y **borra** los commits posteriores. | **Sí** (peligroso) |

**Modo *detached HEAD*.** Cuando usted viaja a un commit antiguo, `HEAD` deja de apuntar a una rama y apunta directamente a ese commit:

```
C0 ── C1 ── C2 ── C3 ── C4 ── C5   ← main
             ↑
            HEAD  (usted está aquí, "de visita" en el pasado)
```

Puede mirar, ejecutar y explorar libremente. Si hace cambios ahí y quiere conservarlos, cree una rama con `git switch -c <nombre>`; si no, simplemente regrese con `git switch main`.

**`revert` vs `reset`.** `git revert` es la forma **segura** de deshacer, porque no borra nada: agrega un commit que invierte los cambios, y la historia en GitHub sigue siendo coherente. `git reset --hard` **reescribe** la historia; si esos commits ya estaban en GitHub, obligará a hacer `git push --force` y puede causar problemas a otros colaboradores. **En este curso use `revert`.**

> **Desde GitHub (sin terminal):** en la página del repositorio haga clic en el contador de **commits** (icono de reloj, *History*). Allí verá la lista de commits; al abrir uno verá sus cambios, y con el botón **Browse files** (`<>`) podrá navegar el repositorio completo tal como estaba en ese momento.
